```python
import pandas as pd
import numpy as np

amazon = spark.table("workspace.amazon_data.amazon").toPandas()


print(amazon.shape)
display(amazon.head())
```

```python
customer_360 = (
    amazon.groupby("CustomerID")
    .agg(
        TotalOrders=("OrderID", "nunique"),
        TotalSpend=("TotalAmount", "sum"),
        AverageOrderValue=("TotalAmount", "mean"),
        FirstOrderDate=("OrderDate", "min"),
        LastOrderDate=("OrderDate", "max"),
        UniqueProducts=("ProductID", "nunique"),
        UniqueCategories=("Category", "nunique"),
        TotalQuantity=("Quantity", "sum")
    )
    .reset_index()
)
```

```python
spark.createDataFrame(customer_360).write.mode(
    "overwrite"
).saveAsTable("amazon_customer_360")
```
